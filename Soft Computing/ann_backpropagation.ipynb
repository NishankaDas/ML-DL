```python
import numpy as np
```

```python
# Sigmoid Function
# -------------------------------------------------

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(y):
    return y * (1 - y)
```

```python
# Inputs
# -------------------------------------------------

x1 = 10
x2 = 5
x3 = 3

# Targets
t6 = 1
t7 = 0

# Learning Rate
eta = 0.01
```

```python
# Input -> Hidden Weights
# -------------------------------------------------

w41 = 0.1
w42 = 0.2
w43 = 0.3

w51 = 0.2
w52 = 0.1
w53 = 0.3
```

```python
# Hidden -> Output Weights
# -------------------------------------------------

w64 = 0.2
w65 = 0.1

w74 = 0.2
w75 = 0.2
```

```python
# =================================================
# TRAINING
# =================================================

epochs = 30000

for epoch in range(1, epochs + 1):
    if epoch%100 == 0:
        print("\n" + "="*60)
        print(f"Epoch {epoch}")
        print("="*60)

    # -------------------------
    # Feed Forward
    # -------------------------

    net4 = x1*w41 + x2*w42 + x3*w43
    o4 = sigmoid(net4)

    net5 = x1*w51 + x2*w52 + x3*w53
    o5 = sigmoid(net5)

    net6 = o4*w64 + o5*w65
    o6 = sigmoid(net6)

    net7 = o4*w74 + o5*w75
    o7 = sigmoid(net7)
    if epoch%100 == 0:
        print("\nFeed Forward")
        print(f"O4 = {o4:.6f}")
        print(f"O5 = {o5:.6f}")
        print(f"O6 = {o6:.6f}")
        print(f"O7 = {o7:.6f}")

    # -------------------------
    # Error
    # -------------------------

    E6 = 0.5*(t6-o6)**2
    E7 = 0.5*(t7-o7)**2
    TotalError = E6 + E7
    if epoch%100 == 0:
        print(f"\nTotal Error = {TotalError:.6f}")

    # -------------------------
    # Backpropagation
    # -------------------------

    delta6 = (t6-o6)*sigmoid_derivative(o6)
    delta7 = (t7-o7)*sigmoid_derivative(o7)

    delta4 = sigmoid_derivative(o4)*(delta6*w64 + delta7*w74)
    delta5 = sigmoid_derivative(o5)*(delta6*w65 + delta7*w75)

    # -------------------------
    # Update Hidden -> Output
    # -------------------------

    w64 += eta*delta6*o4
    w65 += eta*delta6*o5

    w74 += eta*delta7*o4
    w75 += eta*delta7*o5

    # -------------------------
    # Update Input -> Hidden
    # -------------------------

    w41 += eta*delta4*x1
    w42 += eta*delta4*x2
    w43 += eta*delta4*x3

    w51 += eta*delta5*x1
    w52 += eta*delta5*x2
    w53 += eta*delta5*x3
    if epoch%100 == 0:
        print("\nUpdated Weights")

        print(f"w41 = {w41:.6f}")
        print(f"w42 = {w42:.6f}")
        print(f"w43 = {w43:.6f}")

        print(f"w51 = {w51:.6f}")
        print(f"w52 = {w52:.6f}")
        print(f"w53 = {w53:.6f}")

        print(f"w64 = {w64:.6f}")
        print(f"w65 = {w65:.6f}")

        print(f"w74 = {w74:.6f}")
        print(f"w75 = {w75:.6f}")
```
