# Descriptive Implementation of Binary Fuzzy relation

Importing Libraries

```python
import numpy as np
```

Defining Sets

```python
x = [5,8,4,6,2]
y = [9,7,1,3,8]
```

Relation Matrix

```python
mat = np.zeros((len(x),len(y)))
```

```python
mat
```

Membership Function used in this code

$$
\mu_{R} (x,y) =
\begin{cases}
\frac{y-x}{x+y}, & y>x \\
0, & \text{Other Wise }  \\
\end{cases} 
$$

```python
for i in range(len(x)):
    for j in range(len(y)):
        if y[j]>x[i]:
            mat[i][j] = round((abs(y[j]-x[i])/(y[j]+x[i]+2)),3)
```

```python
mat
```

```python
y = [9,7,1,3,8]
z = [6,7,10,9,4]
```

```python
mat2 = np.zeros((len(y),len(z)))
mat2
```

```python
for i in range(len(y)):
    for j in range(len(z)):
        if z[j]>y[i]:
            mat2[i][j] = round((abs(z[j]-y[i])/(y[j]+z[i]+2)),3)
```

```python
mat2
```

Max Min Composition of Fuzzy Relation 

$$R_1 \circ R_2 = \{ (x, z) \mid \max_{y} \min ( \mu_{R_1}(x,y), \mu_{R_2}(y,z) ) \}$$


```python
new = np.zeros((len(x),len(z)))
for i in range(len(mat)):
    for j in range(len(mat[0])):
        if mat[i][j] < mat2[j][i]:
            new[i][j] = mat[i][j]
        else:
            new[i][j] = mat2[j][i]
```
