# Descriptive Implementation of Singel Layer Perceptron 

Three singel layer perceptrons are implemented here, 

* McCulloch-Pitts (MP) neuron
* Hebb neuron model
* Single Layer Perceptron

Importing Library 

```python
import numpy as np
```

McCulloch-Pitts (MP) neuron 

```python
x = np.array([2,5,8])
w = np.array([0.4,0.7,0.1])
```

```python
o = x@w
```

```python
threshold = 4
if o >= threshold:
    print("1")
else:
    print("0")
```

Hebb neuron model

```python
y = 4
eta = 0.1
```

```python
o = x@w
w = w + (eta * x *y)
```

```python
w
```

```python
x@w
```

Singel Layer Perceptron 

```python
x = np.array([2,5,8])
w = np.array([0.4,0.7,0.1])
```

```python
def sig(a):
    return (1/(1+ (2.71)**(-a)))
```

```python
o = x@w
```

```python
w = w + eta*(o-y)*x
```

```python
w
```

```python
sig(x@w)
```
