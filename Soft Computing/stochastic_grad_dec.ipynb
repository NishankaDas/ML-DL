# Descriptive Implementation of Stocastic Gradient Decent 

Importing Libraries

```python
import math
import random
```

Dataset

```python
x = [1,2,3,4,5]
y = [3,5,7,9,11]
w = random.random()
```

SGD

```python
def grad(x,y,w,lr):
    for i in range(len(x)):
        out = x[i]*w
        w = w - lr*(out-y[i])
    return w
```

Results

```python
epochs = 60
for epoch in range(epochs):
    w = grad(x, y, w, lr=0.01)

    pred = [round(i * w, 2) for i in x]
    if (epoch+1)%10 == 0:
        print(f"====== EPOCH {epoch+1} ======")
        print(f"Weight = {w:.4f}")
        print(f"Prediction = {pred}")
        print()
print(f"The original value is {y}")
```
