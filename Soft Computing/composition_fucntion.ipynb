# Descriptive Implementation of Composite function 

Here we are assuming two functions $R$ and $S$ where if the set if obtained as a matirx the points where if there is an realtion between $a_0$ and $b_0$ there will be 1 value at matrix R in position [0,0] and same for S matrix, but with elements $b_1,b_2,b_3$ and $c_1,c_2,c_3$.

Improting library

```python
import numpy as np
```

Relation R

```python
R = [[1,0,0],
     [0,1,1],
     [1,0,0]]
```

Relation S

```python
S = [[0,1,0],
     [0,0,1],
     [0,1,0]]
```

Composite relation $R\odot S = T$ 

```python
T = np.zeros([3,3], dtype=np.uint8).tolist()
```

```python
for i in range(len(R)):
    for j in range(len(R[0])):
        if R[i][j]:
            for k in range(len(S[j])):
                if S[j][k]:
                    T[i][k] = 1
```

```python
for i in T:
    print(i)
```
