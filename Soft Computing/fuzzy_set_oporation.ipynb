# Desriptive Implementation of Fuzzy Set Oporations 

The oporators that are implemented are as follows 
* Complement 
$$1 - \mu_A(x)$$
* Union 
$$max(\mu_A(x), \mu_B(x))$$
* Intersection 
$$min(\mu_A(x), \mu_B(x))$$
* Algebric Sum 
$$\mu_A(x) + \mu_B(x) - \mu_A(x)\mu_B(x)$$
* Algebric Product
$$\mu_A(x) \times \mu_B(x)$$
* Bounded Sum 
$$min(1,\mu_A(x) + \mu_B(x))$$
* Bounded Diffrence 
$$max(0,\mu_A(x) - \mu_B(x))$$

Importing Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

Getting Dataset

```python
age_set = [16,10,8,6,7,23,15,12,12,15,17,19,21,16]
```

```python
len(age_set)
```

```python
A = age_set[:7]
```

```python
A
```

```python
B = age_set[7:]
```

```python
B
```

One Membership function

```python
def mf(age_set, i):
    ma = max(age_set)
    mi = min(age_set)
    return round(((ma-i)/(ma-mi)),2)
```

Second Membership Function

```python
def other_mf(age_set, i):
    ma = max(age_set)
    mi = min(age_set)
    return round(((ma-i)**2/(ma-mi)**2),2)
```

```python
mem1_a = []
for i in A:
    mem1_a.append(mf(age_set,i))
    print(f"Value :{i} with MF: {round(mf(age_set,i),2)}")
```

```python
mem2_a = []
for i in A:
    mem2_a.append(other_mf(age_set,i))
    print(f"Value :{i} with MF: {round(other_mf(age_set,i),2)}")
```

```python
mem_b = []
for i in B:
    mem_b.append(mf(age_set,i))
    print(f"Value :{i} with MF: {round(mf(age_set,i),2)}")
```

Complement of $\mu_A(x)$

```python
comp_a =[]
for i in A:
    print(f"Complement MF value of {i} is {round(1-mf(age_set,i),2)}")
    comp_a.append(round(1-mf(age_set,i),2))
```

```python
plt.scatter(mem1_a, A, color = 'red', label = 'Membership Values')
plt.scatter(comp_a, A, color = 'blue', label = 'Complement of Membership values')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Complement of  Membership Function")
plt.legend()
plt.grid(True)

plt.show()
```

Union 

```python
union_A = []
for i in A:
    union_A.append(max(mf(age_set,i), other_mf(age_set,i)))
```

```python
plt.scatter(mem1_a, A, color = 'red', label = 'Membership Values')
plt.scatter(mem2_a, A, color = 'blue', label = 'Other Membership values')
plt.scatter(union_A, A, color = 'green', label = 'Union')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Complement of  Membership Function")
plt.legend()
plt.grid(True)

plt.show()
```

Intersection

```python
intersection_A = []
for i in A:
    intersection_A.append(min(mf(age_set,i), other_mf(age_set,i)))
```

```python
plt.scatter(mem1_a, A, color = 'red', label = 'Membership Values')
plt.scatter(mem2_a, A, color = 'blue', label = 'Other Membership values')
plt.scatter(intersection_A, A, color = 'green', label = 'Itersection')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Complement of  Membership Function")
plt.legend()
plt.grid(True)

plt.show()
```

Algebric Sum

```python
alg_sum = []
for i in A:
    alg_sum.append(round((mf(age_set,i)+ other_mf(age_set,i) - (mf(age_set,i)* other_mf(age_set,i))),3))
```

```python
plt.scatter(mem1_a, A, color = 'red', label = 'Membership Values')
plt.scatter(mem2_a, A, color = 'blue', label = 'Other Membership values')
plt.scatter(alg_sum, A, color = 'green', label = 'Algebric Sum')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Algebric Sum of  Membership Function")
plt.legend()
plt.grid(True)

plt.show()
```

Algebric Product

```python
alg_prod = []
for i in A:
    alg_prod.append(round(( (mf(age_set,i)* other_mf(age_set,i))),3))
```

```python
plt.scatter(mem1_a, A, color = 'red', label = 'Membership Values')
plt.scatter(mem2_a, A, color = 'blue', label = 'Other Membership values')
plt.scatter(alg_prod, A, color = 'green', label = 'Algebric Product')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Algebric Product of  Membership Function")
plt.legend()
plt.grid(True)

plt.show()
```

Bounded Sum

```python
boundedS = []
for i in A:
    boundedS.append(max(1, round(( (mf(age_set,i) + other_mf(age_set,i))),3)))
```

```python
plt.scatter(mem1_a, A, color = 'red', label = 'Membership Values')
plt.scatter(mem2_a, A, color = 'blue', label = 'Other Membership values')
plt.scatter(boundedS, A, color = 'green', label = 'Bounded Sum')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Bounded Sum of  Membership Function")
plt.legend()
plt.grid(True)

plt.show()
```

Bounded  Diffrence 

```python
boundedD = []
for i in A:
    boundedD.append(min(0, round(( (mf(age_set,i) - other_mf(age_set,i))),3)))
```

```python
plt.scatter(mem1_a, A, color = 'red', label = 'Membership Values')
plt.scatter(mem2_a, A, color = 'blue', label = 'Other Membership values')
plt.scatter(boundedD, A, color = 'green', label = 'Bounded Diffrence')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Bounded Diffrence of  Membership Function")
plt.legend()
plt.grid(True)

plt.show()
```
