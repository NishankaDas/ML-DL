# Descriptive implementation of ANN for XOR gate prediction 

---

Importing Libraries 

---

```python
import numpy as np 
```

 -----------------------------
 XOR Dataset

 -----------------------------

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])
```

 -----------------------------
 Sigmoid Activation

 -----------------------------

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

 -----------------------------
 Initialize Parameters

 -----------------------------

```python
np.random.seed(42)

input_size = 2
hidden_size = 2
output_size = 1

W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))

W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

learning_rate = 0.1
epochs = 10000
```

 -----------------------------
 Training

 -----------------------------

```python
for epoch in range(epochs):

    # Forward Propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(final_input)

    # Compute Error
    error = y - predicted_output

    # Backpropagation
    d_output = error * sigmoid_derivative(predicted_output)

    hidden_error = np.dot(d_output, W2.T)
    d_hidden = hidden_error * sigmoid_derivative(hidden_output)

    # Update Weights and Biases
    W2 += learning_rate * np.dot(hidden_output.T, d_output)
    b2 += learning_rate * np.sum(d_output, axis=0, keepdims=True)

    W1 += learning_rate * np.dot(X.T, d_hidden)
    b1 += learning_rate * np.sum(d_hidden, axis=0, keepdims=True)

    # Print loss every 1000 epochs
    if epoch % 1000 == 0:
        loss = np.mean(error ** 2)
        print(f"Epoch {epoch}, Loss: {loss:.6f}")
```

 -----------------------------
 Testing
 
 -----------------------------

```python
print("\nFinal Predictions:")
print(predicted_output)

print("\nBinary Output:")
print((predicted_output > 0.5).astype(int))
```
