# Descriptive Implementation of Adaline Neural Network

---
Importing Library 

---

```python
import numpy as np
```

-----------------------------
 Generate Random Dataset

 -----------------------------

```python
np.random.seed(42)

# 10 samples with 2 features
X = np.random.rand(10, 2)

# Random target values (-1 or 1)
y = np.random.choice([-1, 1], size=10)
```

 -----------------------------
 Initialize Parameters

 -----------------------------

```python
learning_rate = 0.01
epochs = 20

# Random weights and bias
weights = np.random.randn(2)
bias = np.random.randn()

print("Initial Weights:", weights)
print("Initial Bias:", bias)
```

 -----------------------------
 Training Adaline

 -----------------------------

```python
for epoch in range(epochs):

    total_error = 0

    for i in range(len(X)):

        # Net input
        net = np.dot(X[i], weights) + bias

        # Linear activation
        output = net

        # Error
        error = y[i] - output

        # Weight update (Delta Rule)
        weights += learning_rate * error * X[i]
        bias += learning_rate * error

        total_error += error**2

    mse = total_error / len(X)

    print(f"Epoch {epoch+1:2d} | MSE = {mse:.4f}")

print("\nFinal Weights:", weights)
print("Final Bias:", bias)
```

 -----------------------------
 Testing

 -----------------------------

```python
print("\nPredictions:")

for i in range(len(X)):
    net = np.dot(X[i], weights) + bias

    # Convert linear output to class
    prediction = 1 if net >= 0 else -1

    print(f"Input: {X[i]}  Target: {y[i]}  Predicted: {prediction}")
```
