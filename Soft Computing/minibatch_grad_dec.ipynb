```python
import math
import random
```

```python
x = [1,2,3,4,5,6]
y = [3,5,7,9,11,14]
w = random.random()
```

```python
def grad(x, y, w, lr, batch_size=2):
    n = len(x)

    # Process each mini-batch
    for start in range(0, n, batch_size):
        end = min(start + batch_size, n)

        # Current batch
        x_batch = x[start:end]
        y_batch = y[start:end]

        # Predictions
        out = [i * w for i in x_batch]

        # Gradients
        gr = []
        for i in range(len(x_batch)):
            gr.append((out[i] - y_batch[i]) * x_batch[i])

        # Average gradient of the batch
        avg_g = sum(gr) / len(gr)

        # Update weight
        w = w - lr * avg_g

    return w
```

```python
epochs = 6
for epoch in range(epochs):
    w = grad(x, y, w, lr=0.01)

    pred = [round(i * w, 2) for i in x]

    print(f"====== EPOCH {epoch+1} ======")
    print(f"Weight = {w:.4f}")
    print(f"Prediction = {pred}")
    print()
print(f"The original value is {y}")
```
