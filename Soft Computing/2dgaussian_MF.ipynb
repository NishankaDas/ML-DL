# Descriptive Implementation of 2d Gaussian Membership Function

Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

Defining Gaussian Fucntion

```python
def gaussian_2d(x, y, mean_x, mean_y, sigma_x, sigma_y):
    exponent = -(((x - mean_x) ** 2 / (2 * sigma_x ** 2)) + ((y - mean_y) ** 2 / (2 * sigma_y ** 2)))
    return np.exp(exponent)
```

Values

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
```

```python
mean_x, mean_y = 0.0, 0.0
sigma_x, sigma_y = 1.5, 1.5
```

Defining Z axis

```python
Z = gaussian_2d(X, Y, mean_x, mean_y, sigma_x, sigma_y)
```

Plotting 

```python
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(projection='3d')

surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

ax.set_title('2D Gaussian Membership Function')
ax.set_xlabel('X Variable')
ax.set_ylabel('Y Variable')
ax.set_zlabel('Membership Degree')

fig.colorbar(surf, shrink=0.5, aspect=5)
plt.show()
```
