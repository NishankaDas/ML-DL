Importing Libraries 

```python
import math
import random
```

Dataset 

```python
x = [1,2,3,4,5]
y = [3,5,7,9,11]
w = random.random()
```

Gradient Decent function

```python
def grad(x, y, w, lr):
    # predictions
    out = [i * w for i in x]

    # gradients
    gr = []
    for i in range(len(y)):
        gr.append((out[i] - y[i]) * x[i])

    avg_g = sum(gr) / len(gr)

    # update weight
    w = w - lr * avg_g

    return w
```

Result for $n$ epochs

```python
epochs = 6
for epoch in range(epochs):
    w = grad(x, y, w, lr=0.01)

    pred = [round(i * w, 2) for i in x]

    print(f"====== EPOCH {epoch+1} ======")
    print(f"Weight = {w:.4f}")
    print(f"Prediction = {pred}")
    print()
print(f"The original value is {y}")
```
