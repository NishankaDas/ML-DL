```python
def bilinear(img,new_image):
    new_image[0][0] = img[0][0]
    new_image[0][len(new_image)-1] = img[0][len(img)-1]
    new_image[len(new_image)-1][0] = img[len(img)-1][0]
    new_image[len(new_image)-1][len(new_image)-1] = img[len(img)-1][len(img)-1]

    for i in range(1,len(new_image)-1):
        new_image[0][i] = ((len(new_image)-1-i)/(len(new_image)-1))*(new_image[0,0]) + ((i)/(len(new_image)-1))*(new_image[0,len(new_image)-1])
    for i in range(1,len(new_image)-1):
        new_image[len(new_image)-1][i] = ((len(new_image)-1-i)/(len(new_image)-1))*(new_image[len(new_image)-1,0]) + ((i)/(len(new_image)-1))*(new_image[len(new_image)-1,len(new_image)-1])
    for i in range(1,len(new_image)-1):
        new_image[i][0] = ((len(new_image)-1-i)/(len(new_image)-1))*(new_image[0,0]) + ((i)/(len(new_image)-1))*(new_image[len(new_image)-1,0])
    for i in range(1,len(new_image)-1):
        new_image[i][len(new_image)-1] = ((len(new_image)-1-i)/(len(new_image)-1))*(new_image[0,len(new_image)-1]) + ((i)/(len(new_image)-1))*(new_image[len(new_image)-1,len(new_image)-1])

    for i in range(1,len(new_image)-1):
        for k in range(1,len(new_image)-1):
            new_image[i][k] = ((len(new_image)-1-k)/(len(new_image)-1))*(new_image[i,0]) + ((k)/(len(new_image)-1))*(new_image[i,len(img)-1])

    return new_image
```
