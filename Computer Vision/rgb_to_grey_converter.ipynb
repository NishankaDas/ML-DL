# Descriptive Implementation of RGB to Greyscale Conversion

Methods used are : 
1. Averaging RGB values
2. Selecting any one channel and making its value as greyscale 

Importing Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
import warnings
warnings.filterwarnings('ignore')
```

Reading Images

```python
x = cv2.imread("1.jpg")
x = cv2.cvtColor(x, cv2.COLOR_BGR2RGB)
```

```python
x.shape
```

```python
x = np.array(x)
plt.imshow(x)
plt.title("Grayscale Image")
plt.axis('off')      # Hide x and y axes
plt.show()
```

```python
x[0]
```

```python
avg_gry = []

for i in x:
    gry_row = []
    for j in i:
        gry_row.append(int(sum(j) / 3))
    avg_gry.append(gry_row)
```

```python
plt.imshow(avg_gry, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')      # Hide x and y axes
plt.show()
```
