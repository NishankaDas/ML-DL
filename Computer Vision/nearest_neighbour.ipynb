```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

```python
img = np.array([[0,40],
                [90,255]])
```

```python
new_image = np.zeros((8,8),np.uint8)
```

```python
factor = int(len(new_image)/len(img))
```

```python
for i in range(0,len(new_image)):
    for j in range(0,len(new_image[0])):
        new_image[i][j] = img[int(i/factor)][int(j/factor)]
```
