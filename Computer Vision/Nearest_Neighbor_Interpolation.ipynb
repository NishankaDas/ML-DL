# Descriptive implementation of Nearest Neighbor image interpolation.

---

Importing Libraries

---

```python
import numpy as np
import matplotlib.pyplot as plt
```

---

Assumed Image

---

```python
# Original 2x2 grayscale image
img = np.array([[90, 10],
                [0, 250]], dtype=np.uint8)
```

---

Desired Image Size

---

```python
new_h, new_w = 5, 5

old_h, old_w = img.shape
```

```python
# Create empty output image
upscaled = np.zeros((new_h, new_w), dtype=np.uint8)
```

---

 Nearest Neighbor Interpolation

 ---

```python

for i in range(new_h):
    for j in range(new_w):

        # Map output pixel to input pixel
        src_x = round(i * (old_h - 1) / (new_h - 1))
        src_y = round(j * (old_w - 1) / (new_w - 1))

        upscaled[i, j] = img[src_x, src_y]
```

```python
print("Original Image:")
print(img)

print("\nUpscaled Image:")
print(upscaled)
```

---

 Display

 ---

```python

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray', interpolation='nearest')
plt.title("2×2 Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(upscaled, cmap='gray', interpolation='nearest')
plt.title("5×5 Image")
plt.axis("off")

plt.show()
```
