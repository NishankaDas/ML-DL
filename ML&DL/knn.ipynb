# Descriptive implementation KNN Using Iris Data

Importing Libraries 

```python
import numpy as np
import pandas as pd
from collections import Counter
```

Importing Data

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
df = pd.read_csv(url, header=None, names=column_names)
```

Relabeling Class

```python
species_map = {'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica': 2}
df['species'] = df['species'].map(species_map)
```

```python
X = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].values
y = df['species'].values
```

Mixing Data

```python
np.random.seed(42) 
```

```python
indices = np.arange(len(X))
np.random.shuffle(indices)
```

Deviding the dataset into two parts 

```python
test_size = int(0.2 * len(X))
train_indices = indices[test_size:]
test_indices = indices[:test_size]
```

```python
X_train, y_train = X[train_indices], y[train_indices]
X_test, y_test = X[test_indices], y[test_indices]
```

```python
print(f"Train samples: {len(X_train)}")
print(f"Test samples: {len(X_test)}")
```

```python
def euclidean_distance_4d(point1, point2):
    diff0 = point1[0] - point2[0]
    diff1 = point1[1] - point2[1]
    diff2 = point1[2] - point2[2]
    diff3 = point1[3] - point2[3]
    
    distance = np.sqrt((diff0 ** 2) + (diff1 ** 2) + (diff2 ** 2) + (diff3 ** 2))
    return distance
```

Calculating for one element

```python
def predict_single(x_test_point, X_train, y_train, k=3):
    num_train = len(X_train)
    
    # Two separate arrays: one for distances, one for corresponding training indices
    distances = np.zeros(num_train)
    indices = np.arange(num_train)
    
    # Calculate distance to every training sample
    for i in range(num_train):
        distances[i] = euclidean_distance_4d(x_test_point, X_train[i])
        
    # Sort indices based on calculated distances
    sort_order = np.argsort(distances)
    
    sorted_distances = distances[sort_order]
    sorted_indices = indices[sort_order]
    
    # Get top K nearest neighbors
    k_nearest_indices = sorted_indices[:k]
    k_nearest_labels = y_train[k_nearest_indices]
    
    # Majority vote for classification
    most_common = Counter(k_nearest_labels).most_common(1)
    
    return most_common[0][0]
```

```python
def predict(X_test, X_train, y_train, k=3):
    predictions = []
    for x_test_point in X_test:
        pred = predict_single(x_test_point, X_train, y_train, k=k)
        predictions.append(pred)
    return np.array(predictions)
```

```python
K = 3
y_pred = predict(X_test, X_train, y_train, k=K)
```

```python
accuracy = np.sum(y_pred == y_test) / len(y_test)
print(f"KNN Accuracy (K={K}): {accuracy * 100:.2f}%")
```

```python
reverse_map = {0: 'Setosa', 1: 'Versicolor', 2: 'Virginica'}

df_results = pd.DataFrame({
    'Actual': [reverse_map[val] for val in y_test],
    'Predicted': [reverse_map[val] for val in y_pred]
})

df_results.head(10)
```
