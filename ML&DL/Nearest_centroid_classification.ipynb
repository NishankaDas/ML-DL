```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv("CatsAndDogs.csv")
```

```python
data.head()
```

```python
dog_count = 0
for i in data['Animal']:
    if i == 0:
        dog_count = dog_count+1
```

```python
cat_count = 0
for i in data['Animal']:
    if i == 1:
        cat_count = cat_count+1
```

```python
print(f"Cat Count = {dog_count}")
print(f"Dog Count = {cat_count}")
```

```python
d_central_height = data.loc[data["Animal"] == 0, "Height"].sum()/dog_count
c_central_height = data.loc[data["Animal"] == 1, "Height"].sum()/cat_count
```

```python
d_central_weight = data.loc[data["Animal"] == 0, "Weight"].sum()/dog_count
c_central_weight = data.loc[data["Animal"] == 1, "Weight"].sum()/cat_count
```

```python
d_central_length = data.loc[data["Animal"] == 0, "Length"].sum()/dog_count
c_central_length = data.loc[data["Animal"] == 1, "Length"].sum()/cat_count
```

```python
new_data = [29,6,	30	]
```

```python
dist_d = np.sqrt(((new_data[0]-d_central_height)**2) + ((new_data[1]-d_central_length)**2) + ((new_data[2]-d_central_weight)**2))
dist_c = np.sqrt(((new_data[0]-c_central_height)**2) + ((new_data[1]-c_central_length)**2) + ((new_data[2]-c_central_weight)**2))
```

```python
print(f"Central Values of Class 1 {d_central_height}, {d_central_weight}, {d_central_length}")
print(f"Central Values of Class 2 {c_central_height}, {c_central_weight}, {c_central_length}")
```

```python
if dist_c > dist_d:
    print(dist_c, dist_d)
    print("Class 1 ")
else:
    print(dist_c, dist_d)
    print("Class 0 ")
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

# Class 0
class0 = data[data["Animal"] == 0]

# Class 1
class1 = data[data["Animal"] == 1]

ax.scatter(
    class0["Height"],
    class0["Weight"],
    class0["Length"],
    label="Class 0"
)

ax.scatter(
    class1["Height"],
    class1["Weight"],
    class1["Length"],
    label="Class 1"
)

ax.set_xlabel("Height")
ax.set_ylabel("Weight")
ax.set_zlabel("Length")

ax.set_title("3D Plot of Height, Weight and Length")
ax.legend()

plt.show()
```
